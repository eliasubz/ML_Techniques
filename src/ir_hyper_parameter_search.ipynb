```python

# ✅ 1. Imports and setup

import sys
import time
import json
import numpy as np
import pandas as pd
from pathlib import Path


# Make sure project modules can be imported
sys.path.append("src")

# Project imports
from Parser import Parser
from IBL import IBL
from processing_types import (
    NormalizationStrategy,
    EncodingStrategy,
    MissingValuesNumericStrategy,
    MissingValuesCategoricalStrategy,
    RetentionPolicy,
)
from model_types import Models  # Enum for models
from main_copy import run_experiment_portable 
from itertools import product

# Global constants
BASE_PATH = "../datasetsCBR/datasetsCBR"
RESULTS_PATH = "preliminary_results/"
NUM_SPLITS = 10


# ✅ 2. Define datasets and experimental configuration

datasets = ["pen-based"]#, adult]  # Only two datasets for now
instance_reduction_strategy = "IBL3"  # fixed
distance_metric = "cosine"
voting_strategy = "borda"

# Define preprocessing strategies (Enums)
normalization_strategy = NormalizationStrategy.MEAN_NORMALIZE
encoding_strategy = EncodingStrategy.ONE_HOT_ENCODE
missing_values_numeric_strategy = MissingValuesNumericStrategy.MEDIAN
missing_values_categorical_strategy = MissingValuesCategoricalStrategy.MODE
retention_policy = RetentionPolicy.ALWAYS_RETAIN


# ✅ 3. Hyperparameter search for IB3


# Define the hyperparameter grid
z_scores = [0.674, 1.28]                # 75% CI, 90% CI
min_obs_values = [3, 4, 6]              # Minimum observation thresholds

# Results container
results = []

for dataset_name in datasets:
    for z_val, min_obs in product(z_scores, min_obs_values):
        print(f"\n🚀 Running IBL3 on {dataset_name} | Z={z_val} | min_obs={min_obs}")
        start_time = time.time()

        run_experiment_portable(
            dataset_name=dataset_name,
            model=Models.IR_K_IBL,
            normalization_strategy=normalization_strategy,
            encoding_strategy=encoding_strategy,
            missing_values_numeric_strategy=missing_values_numeric_strategy,
            missing_values_categorical_strategy=missing_values_categorical_strategy,
            k=7 if dataset_name == "adult" else 5,
            distance_metric=distance_metric,
            voting_strategy=voting_strategy,
            svm_kernel=None,
            C=None,
            gamma=None,
            Degree=None,
            retention_policy=retention_policy,
            instance_reduction_strategy=instance_reduction_strategy,
            feature_weighting_strategy=None,
            BASE_PATH=BASE_PATH,
            RESULTS_PATH=RESULTS_PATH,
            # 🆕 Extra IB3-specific hyperparameters
            confidence_z_ibl3=z_val,
            min_observations_ibl3=min_obs,
        )

        elapsed = time.time() - start_time
        print(f"✅ Finished {dataset_name} | Z={z_val} | min_obs={min_obs} in {elapsed:.1f}s\n")
```
